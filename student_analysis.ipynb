```python
import pandas as pd
data={
    "Student":["Aung Aung","Mya Mya","Hla","Su","Ko"],
    "Math":[78,90,65,88,72],
    "English":[85,92,70,81,75],
    "Science":[80,87,68,90,78]
}
df=pd.DataFrame(data)
df
```

```python
df['Average']=df[["Math","English","Science"]].mean(axis=1).round(2)
df
```

```python
df["Result"]=df["Average"].apply(
    lambda x: "Pass" if x>60 else "Fail"
)
df
```

```python
df["Result"].value_counts()
```

```python
subject_average=df[['Math','English','Science']].mean(axis=0).round(2)
subject_average
```

```python
subject_average.idxmax()
```

```python
import matplotlib.pyplot as plt
subject_average.plot(kind="bar")
plt.title("Average Marks by Subjects")
plt.xlabel("Subject")
plt.ylabel("Average Marks")
plt.show()
```

```python
import matplotlib.pyplot as plt
subject_average.plot(
    kind='bar',
    figsize=(8,5),
    ylim=(0,100),
    rot=0
)
plt.title("Average Marks by Subject", fontsize=14)
plt.xlabel("Subject",fontsize=14)
plt.ylabel("Average Marks",fontsize=14)
plt.grid(axis="y", alpha=0.5)
plt.tight_layout()
plt.show()
```
